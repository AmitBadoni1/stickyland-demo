1. Run the cells below.
2. Click the **StickyLand** toolbar button to open the persistent sidebar.
3. Drag a notebook cell by its left gutter into StickyLand, or use **+** to create a floating code or Markdown cell.
4. Edit either the floating cell or its notebook source to see them stay synchronized.
5. Enable auto-run on a floating code cell, then change `days` below to update its output.
6. Drag and resize floating cells to arrange a dashboard.

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt

print("Hello World")
print(f"NumPy {np.__version__} | Pandas {pd.__version__}")
```

```python
rng = np.random.default_rng(42)
traffic = pd.DataFrame({
    "day": pd.date_range("2026-01-01", periods=30),
    "visitors": rng.integers(80, 180, size=30),
    "conversion_rate": rng.uniform(0.08, 0.24, size=30),
})
traffic["signups"] = (traffic.visitors * traffic.conversion_rate).round().astype(int)
traffic.head()
```

## Dashboard controls

Pin this Markdown cell and the next two code cells. The `days` value acts as a lightweight dashboard control: edit it, then run it to drive the auto-run chart and summary.

```python
days = 12  # Try a value from 3 to 30
```

```python
view = traffic.tail(days)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(view.day, view.visitors, marker="o", label="Visitors")
ax.plot(view.day, view.signups, marker="o", label="Signups")
ax.set_title(f"Last {days} days")
ax.grid(alpha=0.25)
ax.legend()
fig.autofmt_xdate()
plt.show()
```

```python
pd.DataFrame({
    "metric": ["Visitors", "Signups", "Conversion"],
    "value": [
        int(view.visitors.sum()),
        int(view.signups.sum()),
        f"{view.signups.sum() / view.visitors.sum():.1%}",
    ],
})
```
